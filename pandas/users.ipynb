```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('sample_users.csv')
df.head()
df.tail()
df.shape
```

```python
df.sample(5)
```

```python
df.columns
```

```python
df.info()
```

```python
#missing values
df.isnull().sum()
```

```python
#create new columns
df.columns

```

```python
df['full_name'] = df['first_name'] + '' + df['last_name']
df['email_domain'] = df['email'].str.split('@').str[-1]
df['website_domain'] = df['web'].str.split('//').str[-1].str.split('/').str[0]
df[['full_name', 'email_domain', 'website_domain']].head()
```

```python
#Select specific columns
selected_columns = df[['full_name', 'email_domain', 'website_domain']]
print(selected_columns.head())
```

```python
#selected filtered data
ca_users = df[df['state'] == 'CA']
ca_users.head()
```

```python
len(ca_users)
```

```python
#sorting of data
sorted_users = df.sort_values(by='last_name')
sorted_users[['first_name', 'last_name']].head()
```

```python
# grouping state and counting users
state_counts = df.groupby('state').size().reset_index(name='user_count')
state_counts
```

```python
#export manipulated data
df.to_csv('manipulated_users.csv', index=False)
print("CSV file exported successfully.")
```

```python
#Marplotib visualization
top_states = state_counts.sort_values(by='user_count', ascending=False). head(10)
plt.figure(figsize=(10,6))
plt.bar(top_states['state'], top_states['user_count'], color='skyblue')
plt.xlabel('State')
plt.ylabel('Number of users')
plt.title('Top 10 states by User Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Pie Chart
top_5_states = state_counts.sort_values(by='user_count', ascending=False). head(5)
plt.figure(figsize=(8,8))
plt.pie(top_5_states['user_count'], labels=top_5_states['state'], autopct='%1.1f%%', startangle=140)
plt.title('Top 5 States by User Count')
plt.axis('equal')
plt.show()
```

```python
#Plot line
plt.figure(figsize=(10,6))
plt.plot(top_states['state'], top_states['user_count'], marker='o', color='skyblue')
plt.xlabel('State')
plt.ylabel('Number of users')
plt.title('Top 10 states by User Count')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#Save charts as images
top_states = state_counts.sort_values(by='user_count', ascending=False). head(10)
plt.figure(figsize=(10,6))
plt.bar(top_states['state'], top_states['user_count'], color='skyblue')
plt.xlabel('State')
plt.ylabel('Number of users')
plt.title('Top 10 states by User Count')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('top_states_bar_chart.png')
plt.show()
print("Bar chart saved as 'top_states_bar_chart.png'.")
```
