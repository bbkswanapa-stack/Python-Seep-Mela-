```python
import matplotlib.pyplot as plt
```

```python
# Line Plot
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
temperatures = [22, 24, 19, 23, 25]
plt.plot(days, temperatures, marker='o', label='Temperature')
plt.title('Temperature over a week')
plt.xlabel('Days')
plt.ylabel('Temperature (C)')
plt.legend()
plt.show()
```

```python
#Bar Chart
products = ["Mobile", "Laptop", "Tablet"]
sales =[150, 200, 100]
plt.bar(products, sales, color=['blue', 'green', 'orange'])
plt.title('Products Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```
