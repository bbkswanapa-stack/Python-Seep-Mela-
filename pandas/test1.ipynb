```python
import pandas as pd
```

```python
# creating series from list
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
series
```

```python
#custom index
marks = pd.Series([90, 80, 70], index= ['maths', 'science', 'english'])
marks
```

```python
print(marks['maths'])
```

```python
#creating series from dictionary
data_dict= {'a':1, 'b':2, 'c': 3}
series_dict = pd.Series(data_dict)
series_dict
```

```python
#creating scalar data -- sabbai data lai eutai value indicate garna lai
series = pd.Series(5, index=['Ram', 'Shyam', 'Gita'])
series
```

```python
#Accesing series data
marks = pd.Series([80,85,90], index= ['Ram', 'Shyam', 'Gita'])
print (marks.iloc[0])  #iloc le chai position use garna lai
print (marks['Gita'])
```

```python
# Series Real Example
stock_prices = pd.Series([157,156,142,124,667], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'])
print(stock_prices.iloc[1])
print(stock_prices['Tuesday'])
```

```python
print(stock_prices.mean())
print(stock_prices.max())
print(stock_prices.min())
print(stock_prices[stock_prices > 150])
```

Data Frame with two- dimensional data with rows and columns.

```python
data = {
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25, 30, 35],
    'City' : ['Kathmandu', 'Bhaktapur', 'Lalitpur']
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df['Name'])
print (df[['Name','City']])
```

```python
#access by position rowwise
df.iloc[0]
```

```python
#filtering dataframe
#And condition
filtered_df = df[(df['Age'] > 28) & (df['City'] == 'Bhaktapur')]
filtered_df
```

```python
#Or condition
filtered_df = df[(df['Age'] > 28) | (df['City'] == 'Bhaktapur')]
filtered_df
```

```python
# Crud Add,Update, Delete
# Adding a new column
df['Salary'] = [50000, 60000, 70000]
print(df)
```

```python
#Updating existing column with condition
df.loc[df['Age']> 30, 'Salary'] = df['Salary'] + 10000
df
```

```python
#Deleting a column
 
df.drop('Salary', axis=1, inplace=True)
print(df)
```

```python
#Deleting a row
df.drop(0, axis=0, inplace=True)
df
```
