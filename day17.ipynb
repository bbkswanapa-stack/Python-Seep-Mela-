```python
#Json String
import json
 
person = {
    "name" : "John", 
    "age" : 30,
    'address' : 'ktm',
    "skills": ['python', 'java', 'c++'],
    "is_programmer" : True,
    'phone' : None
}

person_json = json.dumps(person) #json.dumps le python object lai json string ma convert garna lai
print((person_json), type(person_json))
```

```python
import json
json_string = """
{
    "name": "John",
    "age": 30,
    "address": "ktm",
    "skills": [
    "python",
    "java",
    "c++"
    ],
    "is_programmer": true,
    "phone": null
}
"""

x = json.loads(json_string) #json.loads le dictionary ma convert garna lai
print(x, type(x))
```

```python
import json
 
person = {
    "name" : "John", 
    "age" : 30,
    'address' : 'ktm',
    "skills": ['python', 'java', 'c++'],
    "is_programmer" : True,
    'phone' : None
}

# person_json = json.dumps(person) #json.dumps le python object lai json string ma convert garna lai

# with open("person.json", "w") as f:
#     f.write(person_json) #json string lai file ma write garna lai   #Double steps


with open('person.json', 'w') as f:
    json.dump(person, f) #json string lai file ma write garna lai single step ma  garna lai

with open('person.json', 'w') as f:
    json.dump(person, f, indent=4) #indent le json string lai pretty print garna lai, indent ko value le space ko sankhya dekhaucha
```

```python
with open("person.json", "r") as f:
    data = f.read()
    json_data = json.loads(data) #json.loads le json string lai dictionary ma convert garna lai
    print(json_data, type(json_data))
```

```python

with open("person.json", "r") as f:
    json_data = json.load(f) #json.load le json file lai directly dictionary ma convert garna lai
    print(json_data, type(json_data))
```

```python
import csv

with open("sample_users.csv", "r") as f:
    data = csv.DictReader(f) #csv.DictReader le csv file lai read garna lai, csv.DictReader le csv file ko first line lai header ma mancha ra tyo header ko naam le dictionary ko key banaucha, tesaile i['first_name'] le first_name column ko value dekhaucha ra i['address'] le address column ko value dekhaucha
    for i in data:
        print(i['first_name'], i['address']) #csv file ma first_name ra address column haru chan, tesaile i[first_name] le first_name column ko value dekhaucha ra i[address] le address column ko value dekhaucha
```
