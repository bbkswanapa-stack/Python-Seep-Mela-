```python
us_price = {'milk': 2.05, 'bread': 2.6, 'butter': 3.6, 'mobile': 50, 'television': 1000, 'refrigrator': 700}

#get all prices below 5 in us dollars

#non pythonic way
prices = {}

for k,v in us_price.items():
    if v<5:
        prices.update({k: v})
print(prices)


#pythonic way
prices = {
    k:v
     for k,v in us_price.items() 
     if v < 5
     }
print(prices)
```
