```python
#CRUD
import sqlite3

connection = sqlite3.connect('Students.sqlite3')
terminal = connection.cursor()

#insert query
query = 'Insert into student (id, name, address, college, age, gender, faculty,university) values (10, "test", "address", "college", 10, "m", "bca" , "TU")'


terminal.execute(query)
connection.commit()
print(connection)
```
