```python
import csv

student = {'name': "Ram", 'address': 'ktm', 'age': 22}

with open('student.csv', 'w') as f:
    writer = csv.DictWriter(f, fieldnames=student.keys()) #DictWriter le csv file ma dictionary data write garna lai, fieldnames le csv file ma header define garna lai
    writer.writeheader() #writeheader le csv file ma header write garna lai
    writer.writerow(student) #write a single row as once
```

```python
import csv

students = [
    {"name": "John", "address":"ktm", "age": 20},
    {"name": "Ramesh", "address":"ktm", "age": 22},
    {"name": "Suresh", "address":"pokhara", "age": 24},
    {"name": "Mahesh", "address":"biratnagar", "age": 21},
]

with open('student.csv', 'w', newline='', encoding='utf-8') as f: #newline='' le csv file ma extra blank line aune problem lai solve garna lai, encoding='utf-8' le csv file ma unicode character write garna lai
    writer = csv.DictWriter(f, fieldnames=['name', 'address', 'age']) #or fieldnames=[students[0].keys()] #DictWriter le csv file ma dictionary data write garna lai, fieldnames le csv file ma header define garna lai
    writer.writeheader() 
    writer.writerows(students) #write a bulk row as once
```

Comprehension

```python
ls = [1,2,3,4,5]

output = []
for i in ls:
    output.append(i**2)
print(output) #nonpythonic way
```

```python
output = [i**2 for i in ls]
print(output) #pythonic way
```

```python
age = 17
#ternery operator
is_authorized = "unauthorized" if age < 18 else "authorized"
print(is_authorized)
```

```python
numbers = [1,23,4,3,4,2,2]
output= []
for i in numbers:
    if i%2 == 0:
     output.append("even")
    else:
       output.append("odd")
    
print(output)  #nonpythonic
```

```python
numbers = [1,23,4,3,4,2,2]

output = ["even" if i%2 == 0 else "odd" for i in numbers]
print(output)
```

```python
numbers = [-8, -7, 3, -1, 0, 1, 3, 4, 5, -7, 6, 8, 10]

output = []

for i in numbers:
    if i>0:
        output.append(i)
print(output)

#pythonicway
output = [i for i in numbers if i>0]
print(output)
```

Dictionary Comprehension

```python
us_price = {'milk': 2.05, 'bread': 2.6, 'butter': 2.6}

nepal_price = {}

for k,v in us_price.items():
    nepal_price.update ({k:v * 145})

print(nepal_price)


nepal_price = {k:v * 145 for k,v in us_price.items()}
print(nepal_price)

nepal_price = {i: us_price[i]*145 for i in us_price}
print(nepal_price)
```
